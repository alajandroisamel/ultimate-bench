```python
import subprocess

import pandas as pd
import matplotlib.pyplot as plt

# Cores físicos: a partir de ahí los threads extra son hermanos SMT
cpus = subprocess.run(["lscpu", "-p=CORE"], capture_output=True, text=True).stdout
physical = len({l for l in cpus.splitlines() if not l.startswith("#")})

df = pd.read_csv("scaling.csv")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))
ax1.plot(df["threads"], df["speedup"], marker="o", label="medido")
ax1.plot(df["threads"], df["threads"], "k--", alpha=0.5, label="ideal")
ax1.set_ylabel("Speedup (vs 1 thread)")
ax2.plot(df["threads"], df["efficiency"] * 100, marker="o", label="medido")
ax2.axhline(100, color="k", linestyle="--", alpha=0.5, label="ideal")
ax2.set_ylim(0, 110)
ax2.set_ylabel("Eficiencia (%)")
for ax in (ax1, ax2):
    ax.axvline(physical, color="gray", linestyle=":", label=f"{physical} cores físicos")
    ax.set_xscale("log", base=2)
    ax.set_xticks(df["threads"], labels=df["threads"])
    ax.set_xlabel("Threads (OMP_NUM_THREADS = workers)")
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.show()
```
