```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results.csv")

fig, ax = plt.subplots(figsize=(16, 5))
ax.plot(df["size"], df["ops"] / 1e6, marker="o")
ax.set_xscale("log", base=2)
ax.set_xticks(df["size"], labels=df["size"])
ax.set_ylim(0, df["ops"].max() / 1e6 * 1.1)
ax.set_xlabel("Size")
ax.set_ylabel("Millones de operaciones por segundo")
ax.grid(True, alpha=0.3)
plt.show()
```
